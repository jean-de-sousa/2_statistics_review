#### **Uniform distribution**

#### A uniform distribution is a type of probability distribution in which all outcomes are equally likely within a certain range. Here are some examples:

##### Examples:
- Number on a six sided dice (1/6 Probability)
- Random number Generator (Assume 100 random numbers 1/100 Probability)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import randint
```

```python
np.random.seed(42)
```

##### Generating random numbers between 0 and 1

```python
data = np.random.uniform(size=100)
```

```python
print(data)
```

##### Dice rolls

```python
n_rolls = 777
```

```python
rolls = np.random.randint(1, 7, size=n_rolls)
```

```python
print(rolls)
```

##### Data points

```python
np.mean(rolls)
```

```python
np.var(rolls)
```

```python
np.std(rolls)
```

##### Histogram

```python
plt.hist(rolls, bins = np.arange(1,8) - 0.5, edgecolor='black', rwidth=0.8)
plt.xlabel('Dice value')
plt.ylabel('Frequency')
plt.title('Histogram Dice Rolls')
plt.xticks(np.arange(1,7))
plt.grid(axis='y', linestyle='--', alpha=0.7)
```

##### PMF

```python
dice_distribution = randint(1,7)
```

```python
values = np.arange(1, 7)
```

```python
pmf_scipy = dice_distribution.pmf(values)
```

```python
plt.bar(values, pmf_scipy, width=0.5, edgecolor='black', alpha=0.7)
plt.xlabel('Dice value')
plt.ylabel('Probability')
plt.title('PMF Dice Rolls')
plt.show()
```

##### CDF

```python
cdf_scipy = dice_distribution.cdf(values)
```

```python
plt.step(values, cdf_scipy, where='post', label='CDF', color='b', marker='o')
plt.xlabel('Dice value')
plt.ylabel('Cumulative Probability')
plt.title('CDF Dice Rolls')
plt.grid(True)
plt.tight_layout()
plt.show()
```
