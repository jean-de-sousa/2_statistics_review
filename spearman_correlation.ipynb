# Spearman Rank Correlation Coefficient

The **Spearman correlation coefficient**, denoted by $\rho_s$ or $r_s$, measures the **strength** and **direction** of the **monotonic relationship** between two variables.  
It is based on the **ranks** of the data rather than the raw values.

Its values lie in the interval:

$$
-1 \le \rho_s \le 1
$$

- $\rho_s = 1$: perfect increasing monotonic relationship  
- $\rho_s = -1$: perfect decreasing monotonic relationship  
- $\rho_s = 0$: no monotonic relationship  


## 1. General Definition

Spearman correlation is simply **Pearson correlation applied to the ranks** of $X$ and $Y$:

$$
\rho_s = \rho_{\text{Pearson}}(\operatorname{rank}(X),\, \operatorname{rank}(Y))
$$

Each value is replaced by its position in the sorted order.


## 2. Sample Formula (General Case)

Given paired observations $(x_1, y_1), \ldots, (x_n, y_n)$, convert each variable to ranks and compute:

$$
r_s =
\frac{
\sum_{i=1}^n (R(x_i) - \overline{R_x})(R(y_i) - \overline{R_y})
}{
\sqrt{
\sum_{i=1}^n (R(x_i) - \overline{R_x})^2
\sum_{i=1}^n (R(y_i) - \overline{R_y})^2
}
}
$$


## 3. Simplified Formula (No Ties)

If no values are tied:

$$
r_s =
1 -
\frac{
6 \sum_{i=1}^n d_i^2
}{
n(n^2 - 1)
}
$$

Where:

- $d_i = R(x_i) - R(y_i)$ is the difference between ranks  
- $n$ is the sample size  


## 4. Interpretation

- Positive correlation → as $X$ increases, $Y$ tends to increase  
- Negative correlation → as $X$ increases, $Y$ tends to decrease  
- Near zero → no monotonic relationship  

Spearman can capture **nonlinear monotonic relationships**, making it more flexible than Pearson in many situations.


## 5. Assumptions

- **Monotonic** (not necessarily linear) relationship  
- Works with **ordinal or continuous** data  
- **Robust to outliers** due to rank transformation  
- Does **not require normality**


```python
import numpy as np
import pandas as pd
from scipy import stats
```

##### Manually

```python
hits = np.array([150, 180, 120, 210, 160])
```

```python
rbis = np.array([75, 90, 50, 110, 85])
```

```python
hits_rank = stats.rankdata(hits)
```

```python
print(hits_rank)
```

```python
rbis_rank = stats.rankdata(rbis)
```

```python
print(rbis_rank)
```

```python
d = hits_rank - rbis_rank
```

```python
n = len(hits)
```

```python
d_squared = d ** 2
```

```python
spearman_rank_manual = 1 - (6 * np.sum(d_squared)) / (n * (n**2 -1))
```

```python
print(spearman_rank_manual)
```

##### Scipy example

```python
spearman_rank_scipy, p_value = stats.spearmanr(hits, rbis)
```

```python
print(spearman_rank_scipy)
```

##### Ordinal data

```python
data = {
    'Hours_studied': [10, 15, 5, 18],
    'Grades': ['C', 'A', 'D', 'B']
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
grade_mapping = {'A': 4, 'B': 3, 'C': 2, 'D': 1}
```

```python
df['Grades_Ordinal'] = df['Grades'].map(grade_mapping)
```

```python
df['Hours_Rank'] = df['Hours_studied'].rank()
```

```python
df['Grades_Rank'] = df['Grades_Ordinal'].rank()
```

```python
spearman_rank_pandas = df[['Hours_Rank', 'Grades_Rank']].corr(method='spearman').iloc[0, 1]
```

```python
print(spearman_rank_pandas)
```
