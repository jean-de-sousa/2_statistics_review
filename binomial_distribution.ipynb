#### The binomial distribution represents the probability distribution of a random variable that counts the number of successes in a fixed number of independent trials. Each trial results in either success (with the probability p) or failure (with probability 1-p)

- **Success (p)** (also known as probability at one trial)
  - 2.5% of views turn into subscribers  
    - \( P(X = 1) = 0.025 \)  
    - \( P(X = 0) = 0.975 \)
- **Trials (n)**
  - Video gets 1,000 views
- **Size (numer of trials)**
  - Look at 100 different videos

######  *A random variable is a function that assigns numbers to the outcomes of a random experiment In other words, it transforms uncertain events - like rolling a die or flipping a coin - into numerical values that we can analyze mathematically. 
###### Ex: Let’s say you flip a coin:
###### The **sample space** is:
###### $$S = \{ \text{heads}, \text{tails} \}$$
###### We can define a **random variable \( X \)** as follows:
###### $$X(\text{heads}) = 1, \quad X(\text{tails}) = 0$$
###### So:
###### - If you get **heads**, then \( X = 1 \)
###### - If you get **tails**, then \( X = 0 \)
###### Now, the random outcome (*heads* or *tails*) is represented by **numbers**, allowing us to calculate **means, variances, and probability distributions**.

```python
import numpy as np
from scipy.stats import binom
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)
```

```python
p_success_subs = 0.025
```

```python
n_trials_subs = 1000
```

```python
size_videos = 100
```

```python
numpy_samples = np.random.binomial(n=n_trials_subs, p=p_success_subs, size=size_videos)
```

```python
print(numpy_samples)
```

```python
p_success_coin = 0.5
```

```python
n_trials_coin = 50
```

```python
size_coin_flips = 25
```

```python
scipy_samples = binom.rvs(n=n_trials_coin, p=p_success_coin, size=size_coin_flips)
```

```python
print(scipy_samples)
```

##### PMF: probability of exact occurrences

```python
binom.pmf(n=n_trials_coin, k=25, p=p_success_coin)
```

##### CDF - cumulated probability for less or equal occurrences

```python
binom.cdf(n=n_trials_subs, k=25, p=p_success_subs)
```

```python
plt.figure(figsize=(8,6))
plt.hist(scipy_samples, bins=10, color='#EC7000', edgecolor='black')
plt.title('Histogram coin flips')
plt.xlabel('Success')
plt.ylabel('Frequency')
plt.show()
```

##### Plotting coin flip PMF

```python
x = np.arange(0, n_trials_coin + 1)
pmf = binom.pmf(x, n=n_trials_coin, p=p_success_coin)
```

```python
plt.figure(figsize=(8,6))
plt.stem(x, pmf)
plt.title('PMF')
plt.xlabel('Success')
plt.ylabel('Probability')
```

##### Plotting coin flip CDF

```python
cdf = binom.cdf(x, n=n_trials_coin, p=p_success_coin)
```

```python
plt.figure(figsize=(8,6))
plt.plot(x, cdf, 'b-', lw=2)
plt.title('CDF coin flips')
plt.xlabel('Success')
plt.ylabel('Cumulative Probability')
```
