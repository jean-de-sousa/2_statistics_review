#### The Poisson is a probability distribution that describes the number of events occurring within a fixed interval of time or space

#### Examples:
- Number of subscribers joining a YouTube channel in day
- Website visitors per hour
- Chargebacks per quarter

It deals with counts of events that are non-negative integer values (0, 1, 2...)

Terms to know:
- **Lambda (λ):** Average rate (mean) at which events occur in a given interval.
    - Average rate of subscribers per day. For instance, if you expect to gain 40 subscribers per day, then λ = 40
- **K:** Number events you are calculating the probability for
    -50 subscribers in a day


```python
import numpy as np
from scipy.stats import poisson
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)
```

```python
lambda_value = 5
```

```python
size = 10
```

```python
poisson_samples = np.random.poisson(lam=lambda_value, size=size)
```

```python
print(poisson_samples)
```

```python
poisson_samples_2 = poisson.rvs(mu=lambda_value, size=size)
```

```python
print(poisson_samples_2)
```

```python
print(poisson_samples_2.mean())
```

```python
print(poisson_samples_2.std())
```

##### Poisson PMF single value

```python
k = 3
```

```python
prob = poisson.pmf(k, lambda_value)
```

```python
print(prob)
```

##### PMF for multiple values

```python
prob2 = poisson.pmf(poisson_samples, lambda_value)
```

```python
print(prob2)
```

##### CDF single point

```python
prob_cdf = poisson.cdf(k, lambda_value)
```

```python
print(prob_cdf)
```

##### CDF for multiple points

```python
prob_cdf_2 = poisson.cdf(poisson_samples, lambda_value)
```

```python
print(prob_cdf_2)
```

##### Stem plot

```python
k_values = np.arange(0, 15)
```

```python
pmf_values = poisson.pmf(k_values, lambda_value)
```

```python
plt.stem(k_values, pmf_values)
plt.title("Poisson distribution")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.show()
```

##### Histogram

```python
sample_size = 1000
```

```python
data = np.random.poisson(lambda_value, sample_size)
```

```python
plt.hist(data, bins=np.arange(0,16)-0.5, edgecolor='black', alpha=0.7)
plt.title("Poisson distribution")
plt.xlabel("Number of Events")
plt.ylabel("Count")
plt.xticks(np.arange(0, 15))
plt.show()
```
